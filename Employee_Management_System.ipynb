```python
# Employee Management System

# Employee details storing in variables

employee_name = "jyothi"
employee_id = 101
company_name = "TCS"
department = "IT"
designation = "Python Developer"
monthly_salary = 35000.50
years_of_experience = 2.5
work_location = "Hyderabad"
employee_active_status = True
```

```python

# Printing employee details

print("EMPLOYEE DETAILS")
print("Employee Name:", employee_name)
print("Employee ID:", employee_id)
print("Company Name:", company_name)
print("Department:", department)
print("Designation:", designation)
print("Monthly Salary:", monthly_salary)
print("Years of Experience:", years_of_experience)
print("Work Location:", work_location)
print("Employee Active Status:", employee_active_status)
```

```python
# Printing data types

print("\nDATA TYPES")
print(type(employee_name))
print(type(employee_id))
print(type(company_name))
print(type(department))
print(type(designation))
print(type(monthly_salary))
print(type(years_of_experience))
print(type(work_location))
print(type(employee_active_status))
```

```python
#bonus challenges
# Taking employee details dynamically using input()

employee_name = input("Enter Employee Name: ")

employee_id = int(input("Enter Employee ID: "))

monthly_salary = float(input("Enter Monthly Salary: "))
```

```python
# Calculating yearly salary

yearly_salary = monthly_salary * 12

print("Yearly Salary :", yearly_salary)
```

```python
# Displaying employee card

print("\n==============================")
print("      EMPLOYEE CARD")
print("==============================")

print("Employee Name :", employee_name)
print("Employee ID   :", employee_id)
print("Monthly Salary:", monthly_salary)
print("Yearly Salary :", yearly_salary)

print("==============================")
```
